# Count Total Images Per Split

```python
def count_images(root_path):
    #Recursively count image files under root_path.
    total = 0
    class_counts = {}
    for cls in sorted(os.listdir(root_path)):
        cls_dir = os.path.join(root_path, cls)
        if not os.path.isdir(cls_dir):
            continue
        imgs = [f for f in os.listdir(cls_dir) if f.lower().endswith(('.jpg','.jpeg','.png','.bmp'))]
        class_counts[cls] = len(imgs)
        total += len(imgs)
    return total, class_counts

train_total, train_counts = count_images(train_path)
val_total,   val_counts   = count_images(val_path)
test_total,  test_counts  = count_images(test_path)

print(f"Training images  : {train_total}")
print(f"Validation images: {val_total}")
print(f"Test images      : {test_total}")
print(f"Total images     : {train_total + val_total + test_total}")
```

# Identify and Count Corrupted Images

```python
import os
from PIL import Image

corrupted_images = []

def check_corrupted_in_directory(directory_path, class_names_list):
    """
    Checks for corrupted images in a specified directory, given a list of class names.
    Appends any corrupted image paths to the global 'corrupted_images' list.
    """
    print(f"Checking for corrupted images in: {directory_path}")
    count_corrupted_in_this_dir = 0
    if not os.path.exists(directory_path):
        print(f"Directory '{directory_path}' not found. Skipping.\n")
        return

    for class_name in class_names_list:
        class_path = os.path.join(directory_path, class_name)
        if not os.path.isdir(class_path):
            print(f"Warning: Class directory not found: {class_path}. Skipping.\n")
            continue

        for fname in os.listdir(class_path):
            # Only check image files
            if not fname.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp')):
                continue

            image_path = os.path.join(class_path, fname)
            try:
                with Image.open(image_path) as img:
                    img.verify()  # Verifies that the file is an actual image
            except (IOError, SyntaxError) as e:
                corrupted_images.append(image_path)
                count_corrupted_in_this_dir += 1
                print(f"Corrupted image found: {image_path}. Error: {e}")
    print(f"Found {count_corrupted_in_this_dir} corrupted images in {os.path.basename(directory_path)}.\n")


# --- Define paths and class names explicitly for robustness --- #
# Make sure these paths match your actual Drive setup
base_path = '/content/drive/MyDrive/Dataset'
train_path = os.path.join(base_path, 'train')
val_path = os.path.join(base_path, 'val')     # Sometimes called 'validation'
test_path = os.path.join(base_path, 'test')

# Get class names from the training directory (assuming it's representative)
if os.path.exists(train_dir) and os.path.isdir(train_path):
    class_names = sorted([d for d in os.listdir(train_path) if os.path.isdir(os.path.join(train_path, d))])
    if not class_names:
        print(f"Warning: No class directories found in '{train_path}'. Class names list will be empty.")
else:
    print(f"Error: Training directory '{train_path}' not found or is not a directory. Cannot infer class names.")
    class_names = [] # Ensure class_names is an empty list if directory is invalid


# Perform checks for all directories
check_corrupted_in_directory(train_path, class_names)
check_corrupted_in_directory(val_path, class_names)
check_corrupted_in_directory(test_path, class_names)


print(f"Total corrupted images found across all datasets: {len(corrupted_images)}")
```

# Task 2.5.2 – Design, Train, and Evaluate a Baseline CNN Model

```python
def build_baseline_model(num_classes=num_classes, input_shape=(124, 124, 3)):
    """
    Baseline CNN: 3 Conv-Pool blocks + 3 FCN layers.
    Kernel sizes: 3x3 throughout.
    Filters: 32 -> 64 -> 128 (doubling each block).
    Activation: ReLU (hidden), Softmax (output).
    Optimizer: Adam.
    Loss: Categorical Crossentropy.
    Metrics: Accuracy.
    """
    model = models.Sequential([
        # Block 1
        layers.Conv2D(32, (3, 3), activation='relu', padding='same',
                      input_shape=input_shape, name='conv1'),
        layers.MaxPooling2D((2, 2), name='pool1'),

        # Block 2
        layers.Conv2D(64, (3, 3), activation='relu', padding='same', name='conv2'),
        layers.MaxPooling2D((2, 2), name='pool2'),

        # Block 3
        layers.Conv2D(128, (3, 3), activation='relu', padding='same', name='conv3'),
        layers.MaxPooling2D((2, 2), name='pool3'),

        # Classifier
        layers.Flatten(name='flatten'),
        layers.Dense(256, activation='relu', name='fc1'),
        layers.Dense(128, activation='relu', name='fc2'),
        layers.Dense(num_classes, activation='softmax', name='output')
    ], name='Baseline_CNN')

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

baseline_model = build_baseline_model()
baseline_model.summary()
```

# Train Baseline Mode

```python
EPOCHS_BASELINE = 20

callbacks_baseline = [
    keras.callbacks.EarlyStopping(monitor='val_loss', patience=5,
                                  restore_best_weights=True, verbose=1),
    keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5,
                                      patience=3, min_lr=1e-6, verbose=1)
]

t0 = time.time()
history_baseline = baseline_model.fit(
    train_gen,
    epochs=EPOCHS_BASELINE,
    validation_data=val_gen,
    callbacks=callbacks_baseline,
    verbose=1
)
baseline_train_time = time.time() - t0
print(f"\\nBaseline training time: {baseline_train_time:.1f} seconds")
```

# Baseline Learning Curves

```python
def plot_history(history, title='Model', color='steelblue'):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    ax1.plot(history.history['loss'],     label='Train Loss',      color=color)
    ax1.plot(history.history['val_loss'], label='Val Loss',        color=color, linestyle='--')
    ax1.set_title(f'{title} – Loss Curves', fontweight='bold')
    ax1.set_xlabel('Epoch'); ax1.set_ylabel('Loss')
    ax1.legend(); ax1.grid(alpha=0.3)

    ax2.plot(history.history['accuracy'],     label='Train Acc',   color=color)
    ax2.plot(history.history['val_accuracy'], label='Val Acc',     color=color, linestyle='--')
    ax2.set_title(f'{title} – Accuracy Curves', fontweight='bold')
    ax2.set_xlabel('Epoch'); ax2.set_ylabel('Accuracy')
    ax2.legend(); ax2.grid(alpha=0.3)

    plt.tight_layout()
    fname = title.replace(' ', '_').lower() + '_curves.png'
    plt.savefig(fname, dpi=150)
    plt.show()

plot_history(history_baseline, 'Baseline CNN', 'steelblue')
```

```python
def build_deeper_model(num_classes=NUM_CLASSES, input_shape=(224, 224, 3)):
    """
    Deeper CNN: 6 Conv blocks + BN + Dropout.
    Filters: 64-64-128-128-256-256, doubling the baseline.
    Dropout: 0.5 and 0.3 in FCN head.
    """
    model = models.Sequential([
        # ── Block 1 ───
        layers.Conv2D(64, (3,3), padding='same', input_shape=input_shape, name='d_conv1'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.Conv2D(64, (3,3), padding='same', name='d_conv2'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.MaxPooling2D((2,2)),

        # ── Block 2 ───
        layers.Conv2D(128, (3,3), padding='same', name='d_conv3'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.Conv2D(128, (3,3), padding='same', name='d_conv4'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.MaxPooling2D((2,2)),

        # ── Block 3 ───
        layers.Conv2D(256, (3,3), padding='same', name='d_conv5'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.Conv2D(256, (3,3), padding='same', name='d_conv6'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.MaxPooling2D((2,2)),

        # ── Classifier ──
        layers.Flatten(),
        layers.Dense(512, activation='relu', name='d_fc1'),
        layers.Dropout(0.5),
        layers.Dense(256, activation='relu', name='d_fc2'),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation='softmax', name='d_output')
    ], name='Deeper_CNN')

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

deeper_model = build_deeper_model()
deeper_model.summary()
```

# Train Deeper Model (Adam)

```python
EPOCHS_DEEPER = 30

callbacks_deeper = [
    keras.callbacks.EarlyStopping(monitor='val_loss', patience=7,
                                  restore_best_weights=True, verbose=1),
    keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5,
                                      patience=4, min_lr=1e-7, verbose=1)
]

t0 = time.time()
history_deeper = deeper_model.fit(
    train_gen,
    epochs=EPOCHS_DEEPER,
    validation_data=val_gen,
    callbacks=callbacks_deeper,
    verbose=1
)
deeper_train_time = time.time() - t0
print(f"\\nDeeper model training time (Adam): {deeper_train_time:.1f} seconds")
```

# Ablation – Deeper Model WITHOUT Dropout

```python
def build_no_dropout_model(num_classes=NUM_CLASSES, input_shape=(224, 224, 3)):
    """Identical to deeper model but Dropout layers are removed."""
    model = models.Sequential([
        layers.Conv2D(64, (3,3), padding='same', input_shape=input_shape),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.Conv2D(64, (3,3), padding='same'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.MaxPooling2D((2,2)),

        layers.Conv2D(128, (3,3), padding='same'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.Conv2D(128, (3,3), padding='same'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.MaxPooling2D((2,2)),

        layers.Conv2D(256, (3,3), padding='same'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.Conv2D(256, (3,3), padding='same'),
        layers.BatchNormalization(), layers.Activation('relu'),
        layers.MaxPooling2D((2,2)),

        layers.Flatten(),
        layers.Dense(512, activation='relu'),
        # NO DROPOUT
        layers.Dense(256, activation='relu'),
        # NO DROPOUT
        layers.Dense(num_classes, activation='softmax')
    ], name='Deeper_No_Dropout')

    model.compile(optimizer=keras.optimizers.Adam(1e-3),
                  loss='categorical_crossentropy', metrics=['accuracy'])
    return model

ablation_model = build_no_dropout_model()

callbacks_abl = [
    keras.callbacks.EarlyStopping(monitor='val_loss', patience=5,
                                  restore_best_weights=True, verbose=1)
]

history_abl = ablation_model.fit(
    train_gen, epochs=20, validation_data=val_gen,
    callbacks=callbacks_abl, verbose=1
)

test_gen.reset()
abl_loss, abl_acc = ablation_model.evaluate(test_gen, verbose=0)
print(f"\\nAblation (No Dropout) Test Accuracy: {abl_acc:.4f}")
print(f"Deeper    (With Dropout) Test Accuracy: {deeper_acc:.4f}")
print(f"\\nAccuracy drop from removing Dropout: {deeper_acc - abl_acc:+.4f}")
```

# **Part B: Fine-Tuning a Pre-Trained Model (Transfer Learning)**

# Task 2.6.1 – Loading and Adapting a Pre-Trained Model
## Load MobileNetV2 and Build Transfer Learning Model

```python
from tensorflow.keras.applications import MobileNetV2

# Load pre-trained base (no top layers, ImageNet weights)
base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,       # Remove original 1000-class head
    weights='imagenet'
)
base_model.trainable = False  # Freeze entire base for Phase 1

# Build custom classification head
inputs  = keras.Input(shape=(224, 224, 3))
x = base_model(inputs, training=False)          # Pass frozen base
x = layers.GlobalAveragePooling2D()(x)           # Replace Flatten – reduces params
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)

tl_model = keras.Model(inputs, outputs, name='MobileNetV2_TL')

tl_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

tl_model.summary()
print(f"\\nTotal params      : {tl_model.count_params():,}")
print(f"Trainable params  : {sum([tf.size(v).numpy() for v in tl_model.trainable_variables]):,}")
print(f"Non-trainable     : {sum([tf.size(v).numpy() for v in tl_model.non_trainable_variables]):,}")
```

# Task 2.6.2 – Model Training and Fine-Tuning
## Phase 1 – Feature Extraction (Frozen Base)

```python
EPOCHS_PHASE1 = 15

callbacks_tl_p1 = [
    keras.callbacks.EarlyStopping(monitor='val_loss', patience=5,
                                  restore_best_weights=True, verbose=1)
]

t0 = time.time()
history_tl_phase1 = tl_model.fit(
    train_gen,
    epochs=EPOCHS_PHASE1,
    validation_data=val_gen,
    callbacks=callbacks_tl_p1,
    verbose=1
)
tl_phase1_time = time.time() - t0
print(f"\\nPhase 1 training time: {tl_phase1_time:.1f} seconds")
```

# Phase 2 – Fine-Tuning (Unfreeze Top Layers)

```python
# Unfreeze the top 30 layers of MobileNetV2
base_model.trainable = True
fine_tune_at = len(base_model.layers) - 30

for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

# Recompile with a very low learning rate to avoid catastrophic forgetting
tl_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print(f"Total layers in base: {len(base_model.layers)}")
print(f"Trainable layers after unfreezing: {sum(1 for l in base_model.layers if l.trainable)}")

EPOCHS_PHASE2 = 25

callbacks_tl_p2 = [
    keras.callbacks.EarlyStopping(monitor='val_loss', patience=7,
                                  restore_best_weights=True, verbose=1),
    keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5,
                                      patience=3, min_lr=1e-8, verbose=1)
]

t0 = time.time()
history_tl_phase2 = tl_model.fit(
    train_gen,
    epochs=EPOCHS_PHASE2,
    initial_epoch=history_tl_phase1.epoch[-1] + 1,
    validation_data=val_gen,
    callbacks=callbacks_tl_p2,
    verbose=1
)
tl_phase2_time = time.time() - t0
tl_total_time  = tl_phase1_time + tl_phase2_time
print(f"\\nPhase 2 training time: {tl_phase2_time:.1f} seconds")
print(f"Total TL training time: {tl_total_time:.1f} seconds")
```

# Plot Transfer Learning Curves (Combined Phases)

```python
def concat_histories(h1, h2):
    \"\"\"Concatenate two Keras history dicts across phases.\"\"\"
    combined = {}
    for key in h1.history:
        combined[key] = h1.history[key] + h2.history.get(key, [])
    return combined

tl_combined = concat_histories(history_tl_phase1, history_tl_phase2)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
phase_boundary = len(history_tl_phase1.history['loss'])

for ax, metric, title in [
    (ax1, 'loss',     'Loss'),
    (ax2, 'accuracy', 'Accuracy')
]:
    ax.plot(tl_combined[metric],           label=f'Train {title}', color='seagreen')
    ax.plot(tl_combined[f'val_{metric}'],  label=f'Val {title}',   color='seagreen', linestyle='--')
    ax.axvline(x=phase_boundary, color='grey', linestyle=':', linewidth=1.5, label='Fine-tune starts')
    ax.set_title(f'MobileNetV2 TL – {title}', fontweight='bold')
    ax.set_xlabel('Epoch'); ax.set_ylabel(title)
    ax.legend(); ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('tl_curves.png', dpi=150)
plt.show()
```
